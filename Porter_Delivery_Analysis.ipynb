## Data Cleaning and Preprocessing ##

```python
df.isnull().sum()
```

```python
# Handle Missing Values
df['actual_delivery_time'] = df['actual_delivery_time'].fillna(df['created_at'])
df['store_primary_category'] = df['store_primary_category'].fillna('Unknown')
df['order_protocol'] = df['order_protocol'].fillna('Unknown')
df['total_onshift_partners'] = df['total_onshift_partners'].fillna(0).astype(int)
df['total_busy_partners'] = df['total_busy_partners'].fillna(0).astype(int)
df['total_outstanding_orders'] = df['total_outstanding_orders'].fillna(0).astype(int)
df['market_id'] = df['market_id'].fillna('Unknown')
```

```python
# Convert to datetime
df['created_at'] = pd.to_datetime(df['created_at'])
df['actual_delivery_time'] = pd.to_datetime(df['actual_delivery_time'])
```

```python
# New Features
df['delivery_duration_min'] = (df['actual_delivery_time'] - df['created_at']).dt.total_seconds() / 60
df['day_of_week'] = df['created_at'].dt.day_name()
df['hour_of_day'] = df['created_at'].dt.hour
```

#### Q1. What is the distribution of order volumes across different markets? ####

#### Q6. Analyze the distribution of delivery times. Are there notable outliers? ####

```python
# Analyze delivery time distribution and identify outliers
print(df['delivery_duration_min'].describe())
```

```python
# Visualization
plt.figure(figsize=(10,5))
plt.boxplot(df['delivery_duration_min'].dropna())
plt.title('Distribution of Delivery Times')
plt.ylabel('Delivery Time (Minutes)')
plt.show()
```

```python
# Find outliers using IQR Method
Q1 = df['delivery_duration_min'].quantile(0.25)
Q3 = df['delivery_duration_min'].quantile(0.75)
IQR = Q3-Q1
lower_limit = Q1-1.5 * IQR
upper_limit = Q3+1.5 * IQR
outliers = df[
    (df['delivery_duration_min']<lower_limit) |
    (df['delivery_duration_min']>upper_limit)
    ]
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers:", len(outliers))
```

#### Q8. How do order volumes correlate with the number of busy partners? ####

#### Q3. What is the relationship between the number of on-shift partners and delivery times during peak hours? ####

```python
# Identify peak hours
hourly_orders = df.groupby(df['created_at'].dt.hour).size()
peak_hours = hourly_orders[
    hourly_orders >= hourly_orders.quantile(0.75)
].index
print("Peak Hours:", list(peak_hours))
```

```python
# Analyze delivery time and on-shift partners
peak_data = df[
    df['created_at'].dt.hour.isin(peak_hours)
].copy()
peak_correlation = peak_data[
    ['total_onshift_partners', 'delivery_duration_min']
].corr()
print(peak_correlation)
```

```python
correlation = peak_data[
    'total_onshift_partners'
].corr(
    peak_data['delivery_duration_min']
)
print(
    "Correlation between on-shift partners and delivery time:",
    round(correlation, 3)
)
```

#### Q4. Determine the impact of total items and distinct items in an order on delivery time. ####

```python
# Regression analysis
from sklearn.linear_model import LinearRegression
regression_data = df[
    ['total_items', 'num_distinct_items', 'delivery_duration_min']
].dropna()
X = regression_data[
    ['total_items', 'num_distinct_items']
]
y = regression_data['delivery_duration_min']

model = LinearRegression()

model.fit(X, y)

print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
print("R-squared:", model.score(X, y))
```

```python
coefficients = pd.DataFrame({
    'Feature': ['total_items', 'num_distinct_items'],
    'Coefficient': model.coef_
})

print(coefficients)
```

#### Q9. How does the availability of delivery partners(busy vs on-shift) affect delivery times in different markets? ####

```python
# Busy partners to on-shift partners ratio
df['busy_onshift_ratio'] = (
    df['total_busy_partners'] /
    df['total_onshift_partners'].replace(0, np.nan)
)
market_partner_analysis = (
    df.groupby('market_id')
      .agg(
          avg_busy_onshift_ratio=('busy_onshift_ratio', 'mean'),
          avg_delivery_time=('delivery_duration_min', 'mean'),
          avg_onshift_partners=('total_onshift_partners', 'mean'),
          avg_busy_partners=('total_busy_partners', 'mean')
      )
      .sort_values('avg_delivery_time', ascending=False)
)

print(market_partner_analysis)
```

```python
# Calculate Correlation
correlation = market_partner_analysis[
    'avg_busy_onshift_ratio'
].corr(
    market_partner_analysis['avg_delivery_time']
)
print(
    "Correlation between busy/on-shift ratio and average delivery time:",
    round(correlation, 3)
)
```

```python
# Visualization
plt.figure(figsize=(8, 5))
plt.scatter(
    market_partner_analysis['avg_busy_onshift_ratio'],
    market_partner_analysis['avg_delivery_time']
)

plt.title('Partner Availability vs Average Delivery Time')
plt.xlabel('Busy / On-Shift Partner Ratio')
plt.ylabel('Average Delivery Time (Minutes)')

plt.tight_layout()
plt.show()
```
